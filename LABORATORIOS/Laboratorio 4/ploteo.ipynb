# Ploteo de señales ECG

Código utilizado para cargar y graficar las señales ECG obtenidas con BITalino.

```python
import numpy as np
import matplotlib.pyplot as plt
import glob
import os
```

## Configuración

La frecuencia de muestreo utilizada fue de 1000 Hz. Los archivos `.txt` se encuentran en la carpeta `datos`.

```python
fs = 1000
archivos = glob.glob("datos/*.txt")

print("Archivos encontrados:")
print(archivos)

os.makedirs("ploteos", exist_ok=True)
```

## Procesamiento y ploteo

Se utiliza la columna A1, correspondiente al sexto campo de los archivos, y se construye el eje de tiempo a partir de la frecuencia de muestreo.

```python
for archivo in archivos:
    datos = np.loadtxt(archivo, comments="#")
    ecg = datos[:, 5]
    t = np.arange(len(ecg)) / fs

    nombre = os.path.basename(archivo)
    nombre_sin_extension = os.path.splitext(nombre)[0]

    plt.figure(figsize=(12, 5))
    plt.plot(t, ecg)
    plt.xlabel("Tiempo (s)")
    plt.ylabel("Amplitud")
    plt.title(f"Señal ECG - {nombre_sin_extension}")
    plt.grid()
    plt.tight_layout()

    ruta_guardado = f"ploteos/{nombre_sin_extension}.png"
    plt.savefig(ruta_guardado, dpi=300)
    plt.show()
    plt.close()
    print(f"Guardado: {ruta_guardado}")
```

## Resultado

Las gráficas generadas se guardan automáticamente en la carpeta `ploteos`.
